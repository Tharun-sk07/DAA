```python
import heapq

# -------------------------------
# Union-Find (Disjoint Set)
# -------------------------------
class UnionFind:
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])  # Path Compression
        return self.parent[x]

    def union(self, x, y):
        rootX = self.find(x)
        rootY = self.find(y)

        if rootX == rootY:
            return False

        if self.rank[rootX] < self.rank[rootY]:
            self.parent[rootX] = rootY
        elif self.rank[rootX] > self.rank[rootY]:
            self.parent[rootY] = rootX
        else:
            self.parent[rootY] = rootX
            self.rank[rootX] += 1

        return True


# -------------------------------
# Kruskal's Algorithm
# -------------------------------
def kruskal(vertices, edges):
    uf = UnionFind(vertices)

    edges = sorted(edges, key=lambda x: x[2])

    mst = []
    total_weight = 0

    for u, v, w in edges:
        if uf.union(u, v):
            mst.append((u, v, w))
            total_weight += w

    return mst, total_weight


# -------------------------------
# Prim's Algorithm
# -------------------------------
def prim(vertices, graph):
    visited = [False] * vertices
    min_heap = [(0, 0, -1)]  # (weight, current, parent)

    mst = []
    total_weight = 0

    while min_heap:
        weight, current, parent = heapq.heappop(min_heap)

        if visited[current]:
            continue

        visited[current] = True
        total_weight += weight

        if parent != -1:
            mst.append((parent, current, weight))

        for neighbor, edge_weight in graph[current]:
            if not visited[neighbor]:
                heapq.heappush(min_heap, (edge_weight, neighbor, current))

    return mst, total_weight


# -------------------------------
# Main Program
# -------------------------------
V = 7

# 9 edges (u, v, weight)
edges = [
    (0, 1, 7),
    (0, 3, 5),
    (1, 2, 8),
    (1, 3, 9),
    (1, 4, 7),
    (2, 4, 5),
    (3, 4, 15),
    (3, 5, 6),
    (4, 6, 9)
]

# Build adjacency list
graph = {i: [] for i in range(V)}

for u, v, w in edges:
    graph[u].append((v, w))
    graph[v].append((u, w))

# Kruskal
kruskal_mst, kruskal_weight = kruskal(V, edges)

# Prim
prim_mst, prim_weight = prim(V, graph)

# -------------------------------
# Output
# -------------------------------
print("=" * 50)
print("KRUSKAL'S ALGORITHM")
print("=" * 50)
for u, v, w in kruskal_mst:
    print(f"{u} -- {v}  Weight = {w}")
print("Total MST Weight =", kruskal_weight)

print("\n" + "=" * 50)
print("PRIM'S ALGORITHM")
print("=" * 50)
for u, v, w in prim_mst:
    print(f"{u} -- {v}  Weight = {w}")
print("Total MST Weight =", prim_weight)

print("\n" + "=" * 50)
if kruskal_weight == prim_weight:
    print("Verification Successful!")
    print("Both algorithms produce the same MST weight.")
else:
    print("Verification Failed!")

print("\nTime Complexity Analysis")
print("------------------------")
print("Kruskal's Algorithm : O(E log E)")
print("Prim's Algorithm    : O(E log V) using Min Heap")
```
