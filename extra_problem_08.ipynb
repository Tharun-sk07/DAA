```python
import heapq

def dijkstra_with_hops(graph, source):
    n = len(graph)

    # Distance, hop count, and previous node
    dist = [float('inf')] * n
    hops = [float('inf')] * n
    prev = [None] * n

    dist[source] = 0
    hops[source] = 0

    # Priority Queue: (distance, hops, vertex)
    pq = [(0, 0, source)]

    while pq:
        d, h, u = heapq.heappop(pq)

        if d > dist[u]:
            continue

        for v, w in graph[u]:
            new_dist = d + w
            new_hops = h + 1

            # Update if shorter distance OR same distance with fewer hops
            if (new_dist < dist[v]) or (new_dist == dist[v] and new_hops < hops[v]):
                dist[v] = new_dist
                hops[v] = new_hops
                prev[v] = u
                heapq.heappush(pq, (new_dist, new_hops, v))

    return dist, hops, prev


def reconstruct_path(prev, source, target):
    path = []
    while target is not None:
        path.append(target)
        target = prev[target]

    path.reverse()

    if path and path[0] == source:
        return path
    return []


# Graph with 6 vertices
graph = {
    0: [(1, 2), (2, 4)],
    1: [(2, 1), (3, 7)],
    2: [(3, 3), (4, 2)],
    3: [(5, 1)],
    4: [(3, 1), (5, 5)],
    5: []
}

source = 0

dist, hops, prev = dijkstra_with_hops(graph, source)

print("Shortest Path with Hop Count")
print("-" * 70)
print(f"{'Vertex':<10}{'Distance':<10}{'Hops':<10}{'Path'}")
print("-" * 70)

for v in range(len(graph)):
    path = reconstruct_path(prev, source, v)
    path_str = " -> ".join(map(str, path)) if path else "No Path"

    d = dist[v] if dist[v] != float('inf') else "INF"

    print(f"{v:<10}{d:<10}{hops[v]:<10}{path_str}")
```
