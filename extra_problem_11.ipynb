```python
# Matrix Chain Multiplication using Dynamic Programming

# Matrices:
# T = 4x4
# R = 4x4
# S = 4x4
# V = 4x10000
# P = 10000x1

# Dimensions array
# T(4x4), R(4x4), S(4x4), V(4x10000), P(10000x1)
dims = [4, 4, 4, 4, 10000, 1]

n = len(dims) - 1

# dp[i][j] = minimum scalar multiplications
dp = [[0] * n for _ in range(n)]

# split[i][j] stores the optimal split position
split = [[0] * n for _ in range(n)]

# Chain length
for length in range(2, n + 1):
    for i in range(n - length + 1):
        j = i + length - 1
        dp[i][j] = float('inf')

        for k in range(i, j):
            cost = (dp[i][k] +
                    dp[k + 1][j] +
                    dims[i] * dims[k + 1] * dims[j])

            if cost < dp[i][j]:
                dp[i][j] = cost
                split[i][j] = k


# Function to print optimal order
names = ['T', 'R', 'S', 'V', 'P']

def print_order(i, j):
    if i == j:
        return names[i]

    k = split[i][j]

    left = print_order(i, k)
    right = print_order(k + 1, j)

    return "(" + left + " × " + right + ")"


# Output
print("Optimal Matrix Multiplication Order:")
print(print_order(0, n - 1))

print("\nMinimum number of scalar multiplications:")
print(dp[0][n - 1])
```
