## N-Queens Problem: Standard Backtracking Solution

This section implements the N-Queens problem using a standard backtracking approach with a loop-based `is_safe` function.

```python
import time

def solve_n_queens_standard(n):
    board = [[0 for _ in range(n)] for _ in range(n)]
    solutions = []
    backtracks = {'count': 0}

    def is_safe(board, row, col):
        # Check this row on left side
        for i in range(col):
            if board[row][i] == 1:
                return False

        # Check upper diagonal on left side
        for i, j in zip(range(row, -1, -1), range(col, -1, -1)):
            if board[i][j] == 1:
                return False

        # Check lower diagonal on left side
        for i, j in zip(range(row, n, 1), range(col, -1, -1)):
            if board[i][j] == 1:
                return False

        return True

    def solve_util(board, col):
        if col >= n:
            solutions.append([row[:] for row in board])
            return True

        for i in range(n):
            backtracks['count'] += 1 # Increment backtrack count for each potential placement
            if is_safe(board, i, col):
                board[i][col] = 1
                solve_util(board, col + 1)
                board[i][col] = 0  # Backtrack
        return False # This return is for finding ALL solutions, so it's always False for a given path when backtracking

    solve_util(board, 0)
    return len(solutions), backtracks['count']
```
