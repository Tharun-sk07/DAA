```python
def solve_knight_tour():
    N = 8  # Board size
    board = [[-1 for _ in range(N)] for _ in range(N)]

    # x and y coordinates for the 8 possible moves of a knight
    move_x = [2, 1, -1, -2, -2, -1, 1, 2]
    move_y = [1, 2, 2, 1, -1, -2, -2, -1]

    # Starting position (0,0) and first move number (0)
    board[0][0] = 0

    if not solve_util(N, board, 0, 0, 1, move_x, move_y):
        print("Solution does not exist")
    else:
        print("Solution for Knight's Tour:")
        print_solution(N, board)

def is_safe(N, x, y, board):
    """Checks if (x, y) is a valid and unvisited position on the board."""
    return 0 <= x < N and 0 <= y < N and board[x][y] == -1

def print_solution(N, board):
    """Prints the solution board."""
    for i in range(N):
        for j in range(N):
            print(f"{board[i][j]:>3}", end=" ")
        print()

def solve_util(N, board, curr_x, curr_y, move_count, move_x, move_y):
    """Recursive utility function to find the Knight's Tour."""
    if move_count == N * N:
        return True

    # Try all 8 possible moves from the current coordinate (curr_x, curr_y)
    for i in range(8):
        next_x = curr_x + move_x[i]
        next_y = curr_y + move_y[i]

        if is_safe(N, next_x, next_y, board):
            board[next_x][next_y] = move_count
            if solve_util(N, board, next_x, next_y, move_count + 1, move_x, move_y):
                return True
            # Backtrack: if no solution found from next_x, next_y, reset the move
            board[next_x][next_y] = -1
    return False

# Call the function to solve and display the Knight's Tour
solve_knight_tour()
```
